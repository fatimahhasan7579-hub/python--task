````python
## Task 2: Temperature Converter

### Task Description

#User se Celsius temperature input lein aur usko Fahrenheit mein convert karein. Temperature ka weather message bhi display karein.

### Formula

#```
#Fahrenheit = (Celsius × 9/5) + 32
#```

### Weather Rules

#- Below 10°C: Very Cold
#- 10°C–19°C: Cold
#- 20°C–29°C: Normal
#- 30°C or above: Hot

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9/5) + 32

print("Temperature in Fahrenheit:", fahrenheit)

if celsius < 10:
  print("Very cold")

elif celsius >= 10 and celsius <= 19:
     print("Cold")

elif celsius >= 20 and celsius <= 29:
    print("Normal")

else:
     print("Hot")
````

````python
## Task 3: Mobile Recharge Calculator

### Task Description

#User se recharge amount input lein. Company 3% tax deduct karti hai. Calculate karein ke customer ko kitna final balance milega.

### Formula

#```
#Tax = Recharge Amount × 3 / 100
#Final Balance = Recharge Amount - Tax
#```

recharge_amount = float(input("Enter recharge amount"))
tax = recharge_amount * 3 / 100
final_balance = recharge_amount - tax

print("recharge amount:" , recharge_amount)
print("tax 3%:" , tax)
print("final balance:" , final_balance)
````

````python
## Task 4: Weekly Steps Tracker

### Task Description

#User se seven days ke walking steps input lein. Total aur average steps calculate karein.

### Requirements

#1. `for` loop seven times run karein.
#2. Har din ke steps input lein.
#3. Total steps calculate karein.
#4. Daily average calculate karein.
#5. Agar average `5,000` ya above ho to `Active Week` display karein, warna `Needs More Activity`.

total_step=0

for i in range(7):
  dailystep =int(input("Enter step"))
  total_step=dailystep + total_step
average = total_step/7
if average >= 5000:
  print("Active Week")
else:
  print("Needs More Activity")
````

````python
## Task 5: Secure PIN Checker

### Task Description
#Ek secret PIN program mein save karein. User ko correct PIN enter karne ke liye maximum three attempts dein.

### Requirements

#1. Secret PIN `1234` save karein.
#2. User se PIN input lein.
#3. Correct PIN par `Access Granted` display karein.
#4. Incorrect PIN par remaining attempts show karein.
#5. Three incorrect attempts ke baad `Account Blocked` display karein.

secret_pin = 1234

for attempt in range(3):
    pin = int(input("Enter pin"))

    if pin == secret_pin:
        print("Access Granted")
        break
    else:
        remaining = 2 - attempt
        print("Incorrect pin")
        print("remaining attempts:", remaining)

else:
    print("Account Blocked")
````
