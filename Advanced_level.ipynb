````python
#Task 12: Store Inventory Manager
inventory = {}

while True:
    print("\n1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Product Name: ")
        price = float(input("Price: "))
        qty = int(input("Quantity: "))

        inventory[name] = {
            "price": price,
            "quantity": qty
        }

        print("Product Added!")

    elif choice == "2":
        name = input("Product Name: ")

        if name in inventory:
            qty = int(input("Quantity to Sell: "))

            if inventory[name]["quantity"] >= qty:
                inventory[name]["quantity"] -= qty
                print("Product Sold!")
            else:
                print("Insufficient Stock!")
        else:
            print("Product Not Found!")

    elif choice == "3":
        name = input("Product Name: ")

        if name in inventory:
            qty = int(input("Quantity to Add: "))
            inventory[name]["quantity"] += qty
            print("Stock Updated!")
        else:
            print("Product Not Found!")

    elif choice == "4":
        name = input("Search Product: ")

        if name in inventory:
            print(inventory[name])
        else:
            print("Product Not Found!")

    elif choice == "5":
        total_value = 0

        print("\nInventory:")
        for name, data in inventory.items():
            print(name, data)

            if data["quantity"] < 5:
                print("Low Stock!")

            total_value += data["price"] * data["quantity"]

        print("Total Inventory Value:", total_value)

    elif choice == "6":
        print("Goodbye!")
        break

    else:
        print("Invalid Choice!")
````

````python
#Task 13: Library Book Management System
library = {}

while True:
    print("\n1. Add Book")
    print("2. Issue Book")
    print("3. Return Book")
    print("4. Search Book")
    print("5. Display Books")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        title = input("Book Title: ")
        author = input("Author: ")
        copies = int(input("Copies: "))

        library[title] = {
            "author": author,
            "copies": copies
        }

        print("Book Added!")

    elif choice == "2":
        title = input("Book Title: ")

        if title in library:
            if library[title]["copies"] > 0:
                library[title]["copies"] -= 1
                print("Book Issued!")
            else:
                print("Book Not Available!")
        else:
            print("Book Not Found!")

    elif choice == "3":
        title = input("Book Title: ")

        if title in library:
            library[title]["copies"] += 1
            print("Book Returned!")
        else:
            print("Book Not Found!")

    elif choice == "4":
        title = input("Search Book: ")

        if title in library:
            print(library[title])
        else:
            print("Book Not Found!")

    elif choice == "5":
        print("\nLibrary Books:")
        for title, data in library.items():
            print(title, "-", data["author"], "- Copies:", data["copies"])

    elif choice == "6":
        print("Goodbye!")
        break

    else:
        print("Invalid Choice!")
````

````python

## Task 14: Restaurant Ordering System

### Task Description

#Customer restaurant menu se multiple food items order kar sake aur program complete bill generate kare.
### Example Menu

#```
#menu= {"Burger":500,"Pizza":1200,"Pasta":800,"Drink":200}
#```

### Requirements

#1. Restaurant menu display karein.
#2. Customer se item name aur quantity input lein.
#3. Check karein ke item menu mein available hai.
#4. Multiple items order karne ki facility dein.
#5. Har item ka subtotal calculate karein.
#6. Order ko list of dictionaries mein store karein.
#7. Total bill calculate karein.
#8. Total bill par 13% tax calculate karein.
#9. Rs. 3,000 ya above par 10% discount dein.
#10. Function use karke formatted receipt display karein.

menu={
    "Burger":500,
    "Pizza":1200,
    "Pasta":800,
    "Drink":200,
}
order = []
total = 0
print("====== Resturant menu ======")

while True:
  item_name = input("Enter the item name: ")
  if item_name in menu:
     quantity = int(input("Enter Quantity of your item:"))
     price = menu[item_name] * quantity
     order.append((item_name,quantity,price))
     total = total + price
     print(f"Added {quantity} * {item_name}")

  else:
    print("This item is not  available")

  choice = input("You want to add more items? (yes/no): ")
  if choice.lower() == "no":
     break

tax = total * 0.13
if total >= 3000:
     discount = total * 0.3
else:
    discount = 0

grand_total = total + tax - discount

print(f"Total: {total}")
print(f"Tax: {tax}")
print(f"Discount: {discount}")
print(f"Grand Total: {grand_total}")

def receipt():
     print("====== Resturant menu ======")
     for item in order:
         print(f'order: {item[0]}')
         print(f'quantity: {item[1]}')
         print(f'price: {item[2]}')
         print("-----------------")

receipt()

````

````python
## Task 15: Clinic Appointment Manager

### Task Description

#Aisa program banayein jo clinic ke patients aur appointments manage kare.

### Patient Information

#- Patient ID
#- Patient name
#- Age
#- Disease/problem
#- Appointment time
#- Emergency status

### Requirements
#1. Patient information input lein.
#2. Har patient ko dictionary mein store karein.
#3. Tamam patients ko list mein save karein.
#4. Patient ID duplicate nahi honi chahiye.
#5. Patient ID se record search karein.
#6. Normal aur emergency patients ko identify karein.
#7. Emergency patients ko list mein pehle display karein.
#8. Appointment cancel karne ki facility dein.
#9. Total normal aur emergency patients count karein.
#10. Different operations ke liye separate functions banayein.

### Suggested Menu
#. Add Appointment
#2. Search Patient
#3. Display Appointments
#4. Cancel Appointment
#5. Display Clinic Summary
#6. Exit
patients = []

def add_appointment():
    patient_id = input("Enter Patient ID: ")

    for patient in patients:
        if patient["Patient ID"] == patient_id:
            print("Patient ID already exists!")
            return

    name = input("Enter Patient Name: ")
    age = int(input("Enter Age: "))
    disease = input("Enter Disease/Problem: ")
    appointment_time = input("Enter Appointment Time: ")
    emergency = input("Emergency (Yes/No): ")

    patient = {
        "Patient ID": patient_id,
        "Patient Name": name,
        "Age": age,
        "Disease": disease,
        "Appointment Time": appointment_time,
        "Emergency Status": emergency
    }

    patients.append(patient)
    print("Appointment Added Successfully!")

def search_patient():
    patient_id = input("Enter Patient ID to Search: ")

    for patient in patients:
        if patient["Patient ID"] == patient_id:
            print("\nPatient Found:")
            print(patient)
            return

    print("Patient Not Found!")

def display_appointments():
    print("\n--- Emergency Patients ---")
    for patient in patients:
        if patient["Emergency Status"].lower() == "yes":
            print(patient)

    print("\n--- Normal Patients ---")
    for patient in patients:
        if patient["Emergency Status"].lower() != "yes":
            print(patient)

def cancel_appointment():
    patient_id = input("Enter Patient ID to Cancel: ")

    for patient in patients:
        if patient["Patient ID"] == patient_id:
            patients.remove(patient)
            print("Appointment Cancelled!")
            return

    print("Patient Not Found!")

def clinic_summary():
    emergency_count = 0
    normal_count = 0

    for patient in patients:
        if patient["Emergency Status"].lower() == "yes":
            emergency_count += 1
        else:
            normal_count += 1

    print("\nClinic Summary")
    print("Total Patients:", len(patients))
    print("Emergency Patients:", emergency_count)
    print("Normal Patients:", normal_count)

while True:
    print("\n===== Clinic Appointment Manager =====")
    print("1. Add Appointment")
    print("2. Search Patient")
    print("3. Display Appointments")
    print("4. Cancel Appointment")
    print("5. Display Clinic Summary")
    print("6. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_appointment()
    elif choice == "2":
        search_patient()
    elif choice == "3":
        display_appointments()
    elif choice == "4":
        cancel_appointment()
    elif choice == "5":
        clinic_summary()
    elif choice == "6":
        print("Exiting Program...")
        break
    else:
        print("Invalid Choice!")
````
